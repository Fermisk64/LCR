#### **Question 2:** 
#### Data-visualization

Let's create plots to visualize the relationships between the features (e.g., alcohol, malic acid, ash...).


```python
def plot_feature_pairs(data, feature_names, color_labels=None, title_prefix=''):
    """
    Helper function to create scatter plots for all possible pairs of features.
    
    Parameters:
    - data: DataFrame containing the features to be plotted.
    - feature_names: List of feature names to be used in plotting.
    - color_labels: Optional. Cluster or class labels to color the scatter plots.
    - title_prefix: Optional. Prefix for plot titles to distinguish between different sets of plots.
    """
    # Create a figure for the scatter plots
    plt.figure(figsize=(60, 60))
    
    # Counter for subplot index
    plot_number = 1
    
    # Loop through each pair of features
    for i in range(len(feature_names)):
        for j in range(i + 1, len(feature_names)):
            plt.subplot(len(feature_names)-1, len(feature_names)-1, plot_number)
            
            # Scatter plot colored by labels if provided
            if color_labels is not None:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], 
                            c=color_labels, cmap='viridis', alpha=0.7)
            else:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], alpha=0.7)
            
            plt.xlabel(feature_names[i])
            plt.ylabel(feature_names[j])
            plt.title(f'{title_prefix}{feature_names[i]} vs {feature_names[j]}')
            
            # Increment the plot number
            plot_number += 1

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Show the plot
    plt.show()

# Get feature names
feature_names = wine_df.columns

# Use the helper function to plot scatter plots without coloring by cluster labels
plot_feature_pairs(wine_df, feature_names, title_prefix='Original Data: ')
```

**Question:**
- Do you notice any patterns or relationships between the different features? How might these patterns help in distinguishing between different species?

> Yes, there are clear patterns in the graphs. Alcohol content shows positive correlations with magnesium, proline, total phenols, flavanoids, and color intensity, while ash content rises with alkalinity_of_ash. Strong internal links exist between total phenols and flavanoids, and flavanoids with most features except nonflavanoid phenols. On the negative side, alkalinity_of_ash is inversely related to alcohol, total phenols, flavanoids, proanthocyanins, and hue, and malic acid is negatively related to alcohol, phenolics, and proline. Taken together, wines with higher proline, phenols, flavanoids, and proanthocyanins tend to have higher alcohol and deeper color, reflecting riper, fuller grapes, while higher alkalinity_of_ash and hue correspond to lighter, more acidic wines often linked to younger grapes.

#### **Question 3:** 
#### Data cleaning

```python
# Initialize the StandardScaler
scaler = StandardScaler()

# Scale all the features in the dataset
scaled_features = scaler.fit_transform(wine_df)

# Create a new DataFrame with scaled features
scaled_wine_df = pd.DataFrame(scaled_features, columns=wine_data.feature_names)

# Display the first few rows of the scaled DataFrame
print(scaled_wine_df.head())
```

Why is it important to standardize the features of a dataset before applying clustering algorithms like K-Means? Discuss the implications of using unstandardized data in your analysis.

> Clustering algorithms such as K-mean rely on distance calculations. Unstandardized datasets with unscaled features can distort cluster formation. for instance, variables with larger ranges can dominate the process resulting in distorted and biasd clustering as the result of biased WSSD (Within cluster sum of squared distances). This bias result in clusters that reflect the dominant features rather than whole dataset, and as a result, smaller but important patterns are lost. To summarize, without standardization, it results in reduce interpretability and generalizability of analysis and as aresult, the reiability on the analysis.  

#### **Question 4:** 
#### K-means clustering 

Apply the K-Means clustering algorithm to the Wine dataset. Choose the value 3 for the number of clusters (`k=3`) and fit the model. Assign cluster labels to the original data and add them as a new column in the DataFrame.

```python
# Your answer...

wine_ds_kmeans = KMeans(n_clusters=3)
wine_ds_kmeans.fit(scaled_wine_df)

clustered_wine_data = scaled_wine_df.copy()  # to keep the original copy and be safe
clustered_wine_data["Cluster"] = wine_ds_kmeans.labels_

# checking if the column is added and the results are correct
# print(clustered_wine_data.head())

# Use the helper function to plot scatter plots, colored by cluster labels
plot_feature_pairs(clustered_wine_data, feature_names, color_labels=clustered_wine_data['Cluster'], title_prefix='Clustered Data: ')
```

We chose `k=3` for the number of clusters arbitrarily. However, in a real-world scenario, it is important to determine the optimal number of clusters using appropriate methods.

#### **Question 5:** 
#### Bootstrapping 

Implement bootstrapping on the mean of `color intensity`. Generate 10000 bootstrap samples, calculate the mean for each sample, and compute a 90% confidence interval.

```python
# Your answer here...

mean_color_intensity = [] # we will caalculate the means during the bootstrapping and store the values here

np.random.seed(123)


for i in range(10_000):
    one_sample = clustered_wine_data["color_intensity"].sample(frac=1, replace=True)
    boot_mean = one_sample.mean()
    mean_color_intensity.append(boot_mean)

mean_color_intensity = pd.Series(mean_color_intensity) # since we are storing values 

lower_bound = mean_color_intensity.quantile(0.05)
upper_bound = mean_color_intensity.quantile(0.95)

# Display the result
print(f"Mean of Color Intensity: {mean_color_intensity}")
print(f"90% Confidence Interval of Mean Color Intensity: ({lower_bound}, {upper_bound})")
```

```python
# Visualizing the results to better understand the distribution
plt.figure(figsize=(5, 3))
plt.hist(mean_color_intensity, bins=20, edgecolor='black', alpha=0.7)

# Add titles and labels
plt.title('Distribution of the bootstrap sample means')
plt.xlabel('mean color intensity')
plt.ylabel('frequency')

# Show the plot
plt.grid(True)
plt.show()
```

**Question:**
- Why do we use bootstrapping in this context? What does it help us understand about the mean?

>  In this context, the mean of color_intesnsity only represents the average color intensity of a finite sample or a few of the wines in our dataset. By using the bootstrapping, we can expand our estimate to include a larger number of wines, and as a result, better understanding of the average of color intensity of all of the wines, and a more reliable comparinson between wine groups. for instance, whether the difference in the mean color_intensity is significant or it is just noise. In other words, it helps us understand whether we can rely on the color_intensity as a feature to cluster our wines.

**Question:**
- What is the purpose of calculating the confidence interval from the bootstrap samples? How does it help us interpret the variability of the estimate?

> The purpose is to move beyond a single point estimate of the mean and instead calculate the range of plausible values that the mean could take, based on the variability in our data. It gives us a measure of reliability of the feature. The confidence interval width indicates how variable the means are. For instance, a narrow confidence interval is indicative of less variation in color_intensities and high reliability of the feature. Awide condfidence interval on the other hands, indicates the opposite and there is a siginificant variation and a less reliable feature to conside when clustering data.

**Question:**

- Reflect on the variability observed in the bootstrapped means and discuss whether the mean of the color intensity appears to be a stable and reliable estimate based on the confidence interval and the spread of the bootstrapped means.

> With a confidence interval od approximately 0.25 and a bell‑shaped distribution of means centered around zero, the estimate appears stable and reliable. The symmetric shape indicates small bias, while the clustering of resampled means around zero and the relatively narrow range suggest limited variability in the mean color intensity.
